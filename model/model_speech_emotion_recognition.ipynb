```python
# Cell 8

EMOTIONS = [
    "ANGRY",
    "DISGUSTED",
    "FEARFUL",
    "HAPPY",
    "NEUTRAL",
    "SAD",
    "SURPRISED",
]

EMOTION_TO_ID = {
    emotion: i
    for i, emotion in enumerate(EMOTIONS)
}

ID_TO_EMOTION = {
    i: emotion
    for emotion, i in EMOTION_TO_ID.items()
}

EMOTION_TO_ID
```

```python
# Cell 9

TARGET_SR = 16000
FRAME_SECONDS = 2.0
HOP_SECONDS = 1.0

FRAME_SAMPLES = int(TARGET_SR * FRAME_SECONDS)
HOP_SAMPLES = int(TARGET_SR * HOP_SECONDS)


def load_audio(path, target_sr=TARGET_SR):
    audio, sr = librosa.load(
        path,
        sr=target_sr,
        mono=True
    )

    audio = audio.astype(np.float32)

    # Normalize
    max_val = np.max(np.abs(audio))

    if max_val > 0:
        audio = audio / max_val

    return audio, target_sr


def make_2s_frames(
    audio,
    sr=TARGET_SR,
    frame_seconds=FRAME_SECONDS,
    hop_seconds=HOP_SECONDS,
):
    frame_size = int(sr * frame_seconds)
    hop_size = int(sr * hop_seconds)

    frames = []

    # Pad short audio
    if len(audio) < frame_size:
        padded = np.zeros(frame_size, dtype=np.float32)
        padded[:len(audio)] = audio
        frames.append(padded)
        return frames

    for start in range(
        0,
        len(audio) - frame_size + 1,
        hop_size
    ):
        frame = audio[start:start + frame_size]
        frames.append(frame)

    return frames
```

```python
# Cell 10

def save_temp_frame(
    frame,
    sr=TARGET_SR,
    filename="temp_frame.wav"
):
    sf.write(
        filename,
        frame,
        sr
    )

    return filename
```

```python
# Cell 11

import re


def extract_emotion(text):
    """
    Extract SenseVoice emotion token.
    """

    for emotion in EMOTIONS:
        token = f"<|{emotion}|>"

        if token in text:
            return emotion

    return "NEUTRAL"
```

```python
# Cell 13

def run_sensevoice(
    audio_path,
    model=model,
):
    result = model.generate(
        input=audio_path,
        cache={},
        language="auto",
        use_itn=False,
        batch_size=1,
    )

    return result
```

```python
# Cell 15

def predict_frames(
    audio_path,
    model=model,
    verbose=False,
):
    audio, sr = load_audio(audio_path)

    frames = make_2s_frames(
        audio,
        sr=sr,
    )

    predictions = []

    for i, frame in enumerate(frames):

        temp_path = f"_emotion_frame_{i}.wav"

        save_temp_frame(
            frame,
            sr,
            temp_path,
        )

        try:
            result = run_sensevoice(
                temp_path,
                model=model,
            )

            text = result[0].get("text", "")

            emotion = extract_emotion(text)

            predictions.append({
                "frame": i,
                "emotion": emotion,
                "raw": text,
            })

        finally:
            if os.path.exists(temp_path):
                os.remove(temp_path)

    return predictions
```

```python
# Cell 16

def emotion_percentages(predictions):
    counts = {
        emotion: 0
        for emotion in EMOTIONS
    }

    for prediction in predictions:
        emotion = prediction["emotion"]

        if emotion in counts:
            counts[emotion] += 1

    total = len(predictions)

    if total == 0:
        return {
            emotion: 0.0
            for emotion in EMOTIONS
        }

    percentages = {
        emotion: round(
            100.0 * count / total,
            2
        )
        for emotion, count in counts.items()
    }

    return percentages
```

```python
# Cell 17

def predict_emotion_baseline(
    audio_path,
    model=model,
):
    predictions = predict_frames(
        audio_path,
        model=model,
    )

    percentages = emotion_percentages(
        predictions
    )

    dominant_emotion = max(
        percentages,
        key=percentages.get
    )

    confidence = percentages[dominant_emotion]

    return {
        "emotion": dominant_emotion,
        "confidence": confidence,
        "percentages": percentages,
        "frames": predictions,
    }
```

```python
# Cell 20

def create_2s_dataset(
    input_dir,
    output_dir,
    label,
    language="en",
):
    """
    Convert all audio files in input_dir
    into 2-second training examples.
    """

    input_dir = Path(input_dir)
    output_dir = Path(output_dir)

    output_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    records = []

    audio_extensions = [
        "*.wav",
        "*.mp3",
        "*.flac",
        "*.ogg",
    ]

    files = []

    for extension in audio_extensions:
        files.extend(input_dir.glob(extension))

    for audio_path in tqdm(files):

        audio, sr = load_audio(
            audio_path
        )

        frames = make_2s_frames(
            audio,
            sr=sr
        )

        for frame_id, frame in enumerate(frames):

            output_name = (
                f"{audio_path.stem}"
                f"_frame_{frame_id:04d}.wav"
            )

            output_path = (
                output_dir /
                output_name
            )

            sf.write(
                output_path,
                frame,
                sr
            )

            records.append({
                "key": output_path.stem,
                "source": str(
                    output_path.resolve()
                ),
                "target": "",
                "text_language": f"<|{language}|>",
                "emo_target": f"<|{label}|>",
                "event_target": "<|Speech|>",
                "with_or_wo_itn": "<|woitn|>",
            })

    return records
```
